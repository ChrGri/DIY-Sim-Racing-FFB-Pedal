# Force Fluctuations / Stddev.

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np
from scipy.signal import butter, filtfilt
from scipy.fft import rfft, rfftfreq

ylim = 0.1
# --- Define sampling rate ---
# Calculate fs from your time array if it's not constant
time_diff = np.diff(df["time_InS"])  # convert ms to seconds
fs = 1 / np.median(time_diff)  # effective sampling frequency in Hz

try:
    # --- Filter specifications ---
    cutoff = 200  # Cutoff frequency in Hz
    
    # Sicherstellen, dass mean_sample_time definiert ist
    sampleFreqency = fs 
    if (sampleFreqency < cutoff):
        cutoff = sampleFreqency / 3

    order = 4  # Order of the Butterworth filter
    
    # --- Design low-pass Butterworth filter ---
    b, a = butter(N=order, Wn=cutoff / (0.5 * fs), btype='low')

    # --- Apply zero-phase filtering ---
    df['forceFiltered_offline_InKg'] = filtfilt(b, a, df['pedalForceRaw_fl32'])
    df['forceDeviation'] = df['forceFiltered_offline_InKg'] - df['pedalForceRaw_fl32']

    # Define window duration in seconds
    window_duration_s = 0.05  
    window_samples = int(window_duration_s * fs)

    # Compute rolling std dev
    df['forceDeviation_std'] = (df['forceDeviation']).rolling(window=window_samples, center=True).std()

    # --- BERECHNUNG DES BENCHMARK-WERTES ---
    overall_std_kg = df['forceDeviation'].std()
    overall_std_g = overall_std_kg * 1000  

    # =========================================================================
    # --- DATEN FÜR FREQUENZANALYSE FILTERN (> 0.5 Sekunden) ---
    # Wir erstellen ein neues DataFrame nur für die Frequenzanalyse,
    # damit die Zeit-Plots oben weiterhin den vollen Verlauf zeigen.
    # =========================================================================
    df_fft = df[df["time_InS"] > 0.5].copy()

    # --- FFT BERECHNUNG FÜR SUBPLOT 2 ---
    N = len(df_fft)
    # Wir ziehen den Mittelwert (DC-Anteil) ab, damit der Peak bei 0 Hz das restliche Spektrum nicht optisch erdrückt
    raw_centered = df_fft['pedalForceRaw_fl32'] - df_fft['pedalForceRaw_fl32'].mean()
    dev_centered = df_fft['forceDeviation'] - df_fft['forceDeviation'].mean()

    # Amplitudenspektrum berechnen (Klassische FFT)
    yf_raw = 2.0/N * np.abs(rfft(raw_centered.values))
    yf_dev = 2.0/N * np.abs(rfft(dev_centered.values))
    xf = rfftfreq(N, 1/fs)


    if 0:
        from scipy.interpolate import interp1d

        # 1. Einen perfekten, gleichmäßigen Zeitvektor ohne Lücken erstellen (aus beschnittenen Daten)
        t_uniform = np.arange(df_fft["time_InS"].min(), df_fft["time_InS"].max(), 1/fs)

        # 2. Interpolations-Funktion bauen (linear reicht hier meist völlig aus)
        f_interp_raw = interp1d(df_fft["time_InS"], raw_centered.values, kind='linear')
        f_interp_dev = interp1d(df_fft["time_InS"], dev_centered.values, kind='linear')

        # 3. Das lückenhafte Signal auf das perfekte Raster "umbiegen"
        raw_uniform = f_interp_raw(t_uniform)
        dev_uniform = f_interp_dev(t_uniform)

        # --- Danach machst du deine FFT einfach mit den neuen Daten ---
        N_uniform = len(t_uniform)
        yf_raw = 2.0/N_uniform * np.abs(rfft(raw_uniform))
        yf_dev = 2.0/N_uniform * np.abs(rfft(dev_uniform))
        xf = rfftfreq(N_uniform, 1/fs)

    if 1:
        from scipy.signal import lombscargle

        # Frequenzen definieren, die wir untersuchen wollen (in rad/s, daher * 2 * pi)
        freqs_hz = np.linspace(0.1, 300, 3000) # Von 0.1 Hz bis 300 Hz
        freqs_rad = freqs_hz * 2 * np.pi

        # Lomb-Scargle direkt auf die lückenhaften, beschnittenen Zeit- und Signal-Arrays anwenden
        pgram_raw = lombscargle(df_fft["time_InS"].values, raw_centered.values, freqs_rad, normalize=True)
        pgram_dev = lombscargle(df_fft["time_InS"].values, dev_centered.values, freqs_rad, normalize=True)


    # --- FIGURE LAYOUT AUFBAUEN ---
    # Wir nutzen make_subplots, um eine zweite Achse im oberen Plot zu erlauben (secondary_y)
    fig = make_subplots(
        rows=2, cols=1,
        shared_xaxes=False,
        vertical_spacing=0.12,
        row_heights=[0.7, 0.3], # Der Zeit-Plot bekommt 70% des Platzes, der Frequenz-Plot 30%
        specs=[[{"secondary_y": True}],
               [{"secondary_y": False}]],
        subplot_titles=(f"Time Domain: force with non-causal filter (cutoff: {np.round(cutoff)} Hz)", 
                        "Frequency Domain: Amplitude Spectrum (FFT, computed after 0.5s)")
    )


    # =========================================================================
    # ROW 1: TIME DOMAIN (Primary Y) - Zeigt weiterhin alle Daten aus 'df'
    # =========================================================================
    fig.add_trace(
        go.Scatter(x=df["time_InS"], y=df['pedalForceRaw_fl32'], mode='lines', name='force (raw)'),
        row=1, col=1, secondary_y=False
    )
    fig.add_trace(
        go.Scatter(x=df["time_InS"], y=df['forceFiltered_offline_InKg'], mode='lines', name='force (non-causal filter)'),
        row=1, col=1, secondary_y=False
    )
    if 'pedalForceFiltered_fl32' in df.columns:
        fig.add_trace(
            go.Scatter(x=df["time_InS"], y=df['pedalForceFiltered_fl32'], mode='lines', name='force (filtered)'),
            row=1, col=1, secondary_y=False
        )

    # ROW 1: TIME DOMAIN (Secondary Y)
    fig.add_trace(
        go.Scatter(x=df["time_InS"], y=df['forceDeviation'], mode='lines', name='force (deviation)', opacity=0.8, line=dict(color='#BC8BF8')),
        row=1, col=1, secondary_y=True
    )
    fig.add_trace(
        go.Scatter(x=df["time_InS"], y=df['forceDeviation_std'], mode='lines', name='force deviation std (+)', line=dict(color='orange')),
        row=1, col=1, secondary_y=True
    )
    fig.add_trace(
        go.Scatter(x=df["time_InS"], y=-df['forceDeviation_std'], mode='lines', name='force deviation std (−)', line=dict(color='orange')),
        row=1, col=1, secondary_y=True
    )

    # =========================================================================
    # ROW 2: FREQUENCY DOMAIN (FFT / Lomb-Scargle)
    # =========================================================================
    
    # Lomb-Scargle Traces (basierend auf df_fft)
    fig.add_trace(
        go.Scatter(x=freqs_hz, y=pgram_raw, mode='lines', name='Lomb-Scargle force (raw)', line=dict(color='blue', width=1.5)),
        row=2, col=1
    )
    fig.add_trace(
        go.Scatter(x=freqs_hz, y=pgram_dev, mode='lines', name='Lomb-Scargle force (deviation)', opacity=0.8, line=dict(color='#BC8BF8', width=1.5)),
        row=2, col=1
    )

    # --- TEXTBOX HINZUFÜGEN ---
    fig.add_annotation(
        x=0.01,  
        y=1.05, 
        xref="paper", 
        yref="paper",
        text=f"<b>Benchmark StdDev:</b><br>{overall_std_kg:.4f} kg<br><b>({overall_std_g:.1f} g)</b>",
        showarrow=False,
        font=dict(size=14, color="black"),
        bgcolor="rgba(255, 255, 255, 0.9)",
        bordercolor="black",
        borderwidth=2,
        borderpad=5,
        xanchor="left",
        yanchor="bottom"
    )

    # --- ACHSEN UND LAYOUT FORMATIEREN ---
    fig.update_layout(
        height=900, 
        legend=dict(
            orientation="h",  
            yanchor="bottom",
            y=1.1,  
            xanchor="center",
            x=0.5
        ),
        margin=dict(t=120) 
    )

    # Achsen-Namen für Reihe 1 (Zeit)
    fig.update_xaxes(title_text="time in s", row=1, col=1)
    fig.update_yaxes(title_text="Force in kg", row=1, col=1, secondary_y=False)
    fig.update_yaxes(title_text="Force deviation in kg", range=[-ylim, ylim], row=1, col=1, secondary_y=True)

    # Achsen-Namen für Reihe 2 (Frequenz)
    fig.update_xaxes(title_text="Frequency in Hz", range=[0, 300], row=2, col=1) 
    fig.update_yaxes(title_text="Amplitude", row=2, col=1)

    fig.show()

except Exception as e:
    print(f"Ein Fehler ist aufgetreten: {e}")
```

```python
import numpy as np

def simulate_refined_aom(df, threshold=400.0, attack=30.0, decay=5.0):
    dt = df['timeInUs_u32'].diff().fillna(300) * 1e-6
    force_rate = df['forceVelEst_fl32'].values
    
    intensity = np.zeros(len(df))
    sign_flip_energy = np.zeros(len(df))
    
    curr_intensity = 0.0
    curr_energy = 0.0
    last_sign = 0
    
    for i in range(1, len(df)):
        # 1. Vorzeichenwechsel detektieren
        current_sign = np.sign(force_rate[i])
        if current_sign != last_sign and abs(force_rate[i]) > 50:
            # Energie-Stoß bei jedem Vorzeichenwechsel
            curr_energy += 0.4 
            last_sign = current_sign
        
        # Energie baut sich schnell ab (Zeitkonstante ca. 20ms)
        curr_energy -= 40.0 * dt.iloc[i] 
        curr_energy = max(0, min(1.0, curr_energy))
        sign_flip_energy[i] = curr_energy
        
        # 2. Kombinierte Detektion
        # Nur hochrampen, wenn Rate > Threshold UND genug Wechsel-Energie da ist
        activity = abs(force_rate[i])
        if activity > threshold and curr_energy > 0.3:
            curr_intensity += attack * dt.iloc[i]
        else:
            curr_intensity -= decay * dt.iloc[i]
            
        curr_intensity = max(0, min(1.0, curr_intensity))
        intensity[i] = curr_intensity
        
    return intensity, sign_flip_energy

# Anwendung auf die Daten
df['refinedIntensity'], df['flipEnergy'] = simulate_refined_aom(df)

print(f"AOM Simulation abgeschlossen. Max Intensity im Datensatz: {df['refinedIntensity'].max():.2f}")
```

#Brake Resistor Controller

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np

# --- 1. Tuning Parameters (Prediction) ---
FOOT_ESCAPE_RATE_KG_S = -100.0  
TTZ_WARNING_S = 0.04
MIN_SPEED_HZ = 30000
HOLD_TIME_S = 0.030
ERROR_WAS_LARGE_TRHESHOLD_STEPS_I32 = -100  

# --- 2. Hysteresis Parameters (Reactive Overvoltage Protection) ---
VOLTAGE_THRESHOLD_V = 38.0 
BRAKE_RESISTOR_UPPER_THRESHOLD_VOLTAGE = 4.0  
BRAKE_RESISTOR_LOWER_THRESHOLD_VOLTAGE = 1.0  

# --- 3. Safety Parameters (Thermal Protection) ---
MAX_CONTINUOUS_ON_TIME_S = 0.5  
THERMAL_COOLDOWN_TIME_S = 1.0   

# Arrays extraction (assumes a pandas dataframe 'df' exists)
time_arr = df['time_InS'].values
error_arr = df['servoPositionError_i16'].values
force_rate_arr = df['forceVelEst_fl32'].values
speed_arr = df['currentSpeedInHz_i32'].values
voltage_arr = df['servoVoltage_fl32'].values

# State Variables setup
simulated_brake = np.zeros(len(df))

prev_error = error_arr[0]
is_timer_active = False
timer_start_time_s = 0.0

is_voltage_fallback_active = False

is_hardware_active = False
active_start_time_s = 0.0
is_in_lockout = False
lockout_start_time_s = 0.0

for i in range(1, len(df)):
    t = time_arr[i]
    dt = max(t - time_arr[i-1], 0.000001) 
    
    error = error_arr[i]
    d_error = (error - prev_error) / dt 
    
    force_rate = force_rate_arr[i]
    speed = speed_arr[i]
    voltage = voltage_arr[i]
    
    # --- 1. Time-To-Zero (TTZ) Calculation ---
    ttz = 999.0 
    if (error * d_error) < 0.0:
        ttz = abs(error / d_error)
        
    # --- 2. Foot Dynamics & Kinetic Checks ---
    high_kinetic_energy = abs(speed) > MIN_SPEED_HZ
    foot_is_dynamic = abs(force_rate) > abs(FOOT_ESCAPE_RATE_KG_S)
    error_was_large = (prev_error < ERROR_WAS_LARGE_TRHESHOLD_STEPS_I32) or (error < ERROR_WAS_LARGE_TRHESHOLD_STEPS_I32)
    
    # --- 3. The Precision Trigger ("Sniper Trigger") ---
    trigger = foot_is_dynamic and high_kinetic_energy and (ttz < TTZ_WARNING_S) and error_was_large
        
    # --- 4. Rollover-Safe Timer Logic (NON-RETRIGGERABLE FIX) ---
    if trigger and not is_timer_active:  # <-- HIER IST DIE ANPASSUNG
        is_timer_active = True
        timer_start_time_s = t
        
    prev_error = error
    
    # --- 5. Logical Evaluation ---
    logical_activate = False
    
    if is_timer_active:
        if (t - timer_start_time_s) <= HOLD_TIME_S:
            logical_activate = True
        else:
            is_timer_active = False # Timer ist exakt nach 30ms fertig
            
    # 5b. Reactive Hysteresis
    upper_limit_V = VOLTAGE_THRESHOLD_V + BRAKE_RESISTOR_UPPER_THRESHOLD_VOLTAGE
    lower_limit_V = VOLTAGE_THRESHOLD_V + BRAKE_RESISTOR_LOWER_THRESHOLD_VOLTAGE
    
    if voltage >= upper_limit_V:
        is_voltage_fallback_active = True
    elif voltage <= lower_limit_V:
        is_voltage_fallback_active = False
        
    if is_voltage_fallback_active:
        logical_activate = True
        
    # =========================================================
    # 6. THERMAL OVERLOAD PROTECTION
    # =========================================================
    if is_in_lockout:
        if (t - lockout_start_time_s) > THERMAL_COOLDOWN_TIME_S:
            is_in_lockout = False
        else:
            logical_activate = False 
            
    if logical_activate:
        if not is_hardware_active:
            is_hardware_active = True
            active_start_time_s = t
        else:
            if (t - active_start_time_s) > MAX_CONTINUOUS_ON_TIME_S:
                is_hardware_active = False
                logical_activate = False      
                is_in_lockout = True          
                lockout_start_time_s = t
                is_voltage_fallback_active = False 
                is_timer_active = False 
    else:
        is_hardware_active = False

    if logical_activate:
        simulated_brake[i] = 1.0
        
df['simulated_brake_b'] = simulated_brake * 255
```

```python
import os
import pandas as pd
import numpy as np
from scipy.signal import correlate, correlation_lags
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==============================================================================
# 1. LOGIC ANALYZER DATEN LADEN & PARSEN (NUR WENN VORHANDEN)
# ==============================================================================
LOGIC_CSV_PATH = 'digital.csv'
logic_data_available = os.path.exists(LOGIC_CSV_PATH)

time_shift = 0.0  # Fallback-Wert, falls die Datei nicht existiert

if logic_data_available:
    # FIX 1: Richtung umkehren (vorher falsche Laufrichtung)
    DIR_INVERTED = True  

    df_logic = pd.read_csv(LOGIC_CSV_PATH)
    df_logic.columns = ['Time_s', 'Step', 'Dir']

    # Flankenerkennung (Rising Edge)
    rising_edges = (df_logic['Step'] == 1) & (df_logic['Step'].shift(1) == 0)
    df_steps = df_logic[rising_edges].copy()

    # Richtung auswerten
    if DIR_INVERTED:
        df_steps['Step_Delta'] = df_steps['Dir'].apply(lambda x: -1 if x == 1 else 1)
    else:
        df_steps['Step_Delta'] = df_steps['Dir'].apply(lambda x: 1 if x == 1 else -1)

    # Kumulierte Schritte (Relative Form)
    df_steps['Step_Position_Relative'] = df_steps['Step_Delta'].cumsum()

    # ==========================================
    # NEU: Mathematisch korrekte Geschwindigkeitsberechnung
    # ==========================================
    # Wir betrachten ein Fenster von X Schritten und berechnen Delta Weg / Delta Zeit
    WINDOW = 20

    # Verstrichene Zeit für die letzten N Schritte
    df_steps['Time_Diff'] = df_steps['Time_s'].diff(periods=WINDOW)

    # Zurückgelegte (gerichtete!) Distanz in den letzten N Schritte
    df_steps['Pos_Diff'] = df_steps['Step_Position_Relative'].diff(periods=WINDOW)

    # Frequenz berechnen: Delta Position / Delta Zeit
    # Wir nutzen shift(), um den Datenpunkt in die Mitte des Fensters zurückzuschieben,
    # damit es keinen zeitlichen Versatz (Lag) im Plot gibt.
    df_steps['Speed_Hz'] = (df_steps['Pos_Diff'] / df_steps['Time_Diff']).shift(-WINDOW // 2)


    # ==============================================================================
    # 2. KREUZKORRELATION FÜR ZEIT-SYNCHRONISATION (Time-Alignment)
    # ==============================================================================
    # Beide Signale auf ein 1ms-Raster bringen, um sie vergleichen zu können
    dt_sync = 0.001 
    t_start = min(df['time_InS'].min(), df_steps['Time_s'].min())
    t_end = max(df['time_InS'].max(), df_steps['Time_s'].max())
    t_grid = np.arange(t_start, t_end, dt_sync)

    # Interpolation der relativen Signale
    tel_interp = np.interp(t_grid, df['time_InS'], df['targetPosition_i32'])
    log_interp = np.interp(t_grid, df_steps['Time_s'], df_steps['Step_Position_Relative'])

    # DC-Offset entfernen, sonst schlägt die Korrelation fehl
    tel_norm = tel_interp - np.mean(tel_interp)
    log_norm = log_interp - np.mean(log_interp)  

    # Kreuzkorrelation berechnen
    corr = correlate(tel_norm, log_norm, mode='full')
    lags = correlation_lags(len(tel_norm), len(log_norm), mode='full')
    best_lag = lags[np.argmax(corr)]

    # Zeit-Offset anwenden
    time_shift = best_lag * dt_sync
    df_steps['Time_s_Sync'] = df_steps['Time_s'] + time_shift


    # ==============================================================================
    # 3. Y-ACHSEN OFFSET KORRIGIEREN
    # ==============================================================================
    # Wir holen uns den exakten Telemetrie-Positionswert zum Startzeitpunkt des (nun verschobenen) Logic Analyzers
    start_time_sync = df_steps['Time_s_Sync'].iloc[0]
    y_start_tel = np.interp(start_time_sync, df['time_InS'], df['targetPosition_i32'])

    # Y-Offset so anpassen, dass das Logic-Signal am Start exakt auf der Telemetrie liegt
    df_steps['Step_Position'] = df_steps['Step_Position_Relative'] - df_steps['Step_Position_Relative'].iloc[0] + y_start_tel


# ==============================================================================
# 4. PLOTLY SUBPLOTS
# ==============================================================================
# Titel dynamisch anpassen
title_pos_str = f"1. Position (Time Shift: {time_shift:.3f}s)" if logic_data_available else "1. Position"

fig = make_subplots(
    rows=3, cols=1,
    shared_xaxes=True,
    vertical_spacing=0.08,
    subplot_titles=(
        title_pos_str, 
        "2. Velocity", 
        "3. Acceleration"
    ),
    specs=[[{"secondary_y": True}], [{"secondary_y": True}], [{"secondary_y": False}]]
)

# --- ROW 1: POSITION ---
fig.add_trace(go.Scatter(x=df['time_InS'], y=df['admittance_virtualPosition_m'], mode='lines', name='Virtual Pos [m]', line=dict(color='#1f77b4', width=2)), row=1, col=1, secondary_y=False)

STEPS_PER_METER = 3750 / 0.01
yVirtualSteps = df['admittance_virtualPosition_m'] * STEPS_PER_METER
yVirtualSteps += (df['targetPosition_i32'].iloc[0] - yVirtualSteps.iloc[0]) 

fig.add_trace(go.Scatter(x=df['time_InS'], y=yVirtualSteps, mode='lines', name='Virtual Pos converted [Steps]', line=dict(color='#2ca02c', width=2, dash='dash')), row=1, col=1, secondary_y=True)
fig.add_trace(go.Scatter(x=df['time_InS'], y=df['targetPosition_i32'], mode='lines', name='Target Pos [Steps]', line=dict(color='#ff7f0e', width=2, dash='dot')), row=1, col=1, secondary_y=True)

# NEU: Logic Analyzer Plot wird nur hinzugefügt, wenn Daten vorhanden sind
if logic_data_available:
    fig.add_trace(go.Scatter(x=df_steps['Time_s_Sync'], y=df_steps['Step_Position'], mode='lines', name='Logic Analyzer Pos (Synced)', line=dict(color='#d62728', width=2, dash='solid')), row=1, col=1, secondary_y=True)

# --- ROW 2: VELOCITY ---
fig.add_trace(go.Scatter(x=df['time_InS'], y=df['admittance_virtualVelocity_mps'], mode='lines', name='Virtual Vel [m/s]', line=dict(color='#1f77b4', width=2)), row=2, col=1, secondary_y=False)
fig.add_trace(go.Scatter(x=df['time_InS'], y=df['currentSpeedInHz_i32'], mode='lines', name='Hardware Target Speed [Hz]', line=dict(color='#ff7f0e', width=2, dash='dot')), row=2, col=1, secondary_y=True)

if logic_data_available:
    fig.add_trace(go.Scatter(x=df_steps['Time_s_Sync'], y=df_steps['Speed_Hz'], mode='lines', name='Logic Analyzer Speed (Synced)', line=dict(color='#d62728', width=2, dash='solid')), row=2, col=1, secondary_y=True)

# --- ROW 3: ACCELERATION ---
fig.add_trace(go.Scatter(x=df['time_InS'], y=df['admittance_virtualAcceleration_mps2'], mode='lines', name='Virtual Acc [m/s²]', line=dict(color='#2ca02c', width=2)), row=3, col=1, secondary_y=False)

# --- LAYOUT ---
fig.update_layout(
    title_text="Admittance Model Telemetry vs. Hardware Output",
    height=900,
    showlegend=True, 
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
    hovermode="x unified" 
)

fig.update_yaxes(title_text="Position [m]", title_font=dict(color="#1f77b4"), tickfont=dict(color="#1f77b4"), row=1, col=1, secondary_y=False)
fig.update_yaxes(title_text="Target [Steps]", title_font=dict(color="#ff7f0e"), tickfont=dict(color="#ff7f0e"), row=1, col=1, secondary_y=True)
fig.update_yaxes(title_text="Velocity [m/s]", title_font=dict(color="#1f77b4"), tickfont=dict(color="#1f77b4"), row=2, col=1, secondary_y=False)
fig.update_yaxes(title_text="Speed [Hz]", title_font=dict(color="#ff7f0e"), tickfont=dict(color="#ff7f0e"), row=2, col=1, secondary_y=True)
fig.update_yaxes(title_text="Acceleration [m/s²]", title_font=dict(color="#2ca02c"), tickfont=dict(color="#2ca02c"), row=3, col=1)
fig.update_xaxes(title_text="Time [s]", row=3, col=1)

fig.show()
```
